```python
# Definindo a constante de Hubble (valor aproximado em km/s/Mpc)
H0 = 70 

# Distância da galáxia em Megaparsecs (Mpc)
distancia = 15 

# Calculando a velocidade (v = H0 * d)
velocidade = H0 * distancia

# Mostrando o resultado na tela
print(f"A velocidade de recessão da galáxia é de {velocidade} km/s")
```
